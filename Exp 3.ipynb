```python
import numpy as np
import pandas as pd

# Read the CSV file
df = pd.read_csv("Hotel_Dataset.csv")

# Display the dataset
df

# Identify duplicate rows
df.duplicated()

# Display information about the dataset
df.info()

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Check number of rows
len(df)

# Reset the index
index = np.array(list(range(0, len(df))))
df.set_index(index, inplace=True)

# Drop the extra Age Group column
if 'Age_Group_1' in df.columns:
    df.drop('Age_Group_1', axis=1, inplace=True)
elif 'Age_Group.1' in df.columns:
    df.drop('Age_Group.1', axis=1, inplace=True)

# Replace inappropriate CustomerID values with NaN
df.loc[df['CustomerID'] < 0, 'CustomerID'] = np.nan

# Replace inappropriate Bill values with NaN
df.loc[df['Bill'] < 0, 'Bill'] = np.nan

# Replace inappropriate EstimatedSalary values with NaN
df.loc[df['EstimatedSalary'] < 0, 'EstimatedSalary'] = np.nan

# Replace inappropriate NoOfPax values with NaN
df.loc[
    (df['NoOfPax'] < 1) | (df['NoOfPax'] > 20),
    'NoOfPax'
] = np.nan

# Replace inappropriate Rating values with NaN
df.loc[
    (df['Rating(1-5)'] < 1) | (df['Rating(1-5)'] > 5),
    'Rating(1-5)'
] = np.nan

# Check unique values
df['Age_Group'].unique()

df['Hotel'].unique()

# Correct hotel name
df['Hotel'] = df['Hotel'].replace('Ibys', 'Ibis')

# Check food preference
df['FoodPreference'].unique()

# Correct food preference values
df['FoodPreference'] = df['FoodPreference'].replace(
    ['Vegetarian', 'veg', 'Veg'],
    'Veg'
)

df['FoodPreference'] = df['FoodPreference'].replace(
    ['non-Veg'],
    'Non-Veg'
)

# Fill missing values
# Continuous data -> mean
df['EstimatedSalary'] = df['EstimatedSalary'].fillna(
    round(df['EstimatedSalary'].mean())
)

# Discrete data -> median
df['NoOfPax'] = df['NoOfPax'].fillna(
    round(df['NoOfPax'].median())
)

df['Rating(1-5)'] = df['Rating(1-5)'].fillna(
    round(df['Rating(1-5)'].median())
)

# Continuous data -> mean
df['Bill'] = df['Bill'].fillna(
    round(df['Bill'].mean())
)

# Display final cleaned dataset
df
```
